```python
data=np.array([10,20,30,40,50])
normalized=(data - data.min())/(data.max()-data.min())
print(normalized)
```

```python
import pandas as pd
df=pd.DataFrame({"order_id":[101,102,103,104,105],
                 "customer":["A","B","C","D","E"],
                 "amount":[500,800,300,900,400],
                 "status":["completed","pending","completed","pending","completed"]})
completed=df[df["status"]=="completed"]
print(completed)
print("Total_revenue:",completed["amount"].sum())
print(completed.nlargest(3,"amount"))
```

```python
df=pd.DataFrame({"employee":["A","B","C","D"],
                 "department":["HR","IT","HR","IT"],
                 "salary":[30000,50000,20000,70000]})
avg_salary=df.groupby("department")["salary"].mean()
print(avg_salary)
print("Department with highest average salary:")
print(avg_salary.idxmax())
```

```python
import numpy as np
import pandas as pd
df=pd.DataFrame({"marks":[80,np.nan,70,np.nan,90]})
print(df.isnull())
df["marks"].fillna(df["marks"].mean(),inplace=True)
print(df)
```

```python
import pandas as pd
df = pd.DataFrame({
    "timestamp":["2024-01-10","2024-05-20","2025-03-15"],
    "log_message":["ERROR: File missing",
                   "Success",
                   "ERROR: Connection failed"]})
df["timestamp"] = pd.to_datetime(df["timestamp"])
df["year"] = df["timestamp"].dt.year
errors = df[df["log_message"].str.contains("ERROR")]
print(errors)
print(errors.groupby("year").size())
```

```python
import pandas as pd
customers = pd.DataFrame({
    "customer_id":[1,2,3],
    "customer_name":["A","B","C"]
})
orders = pd.DataFrame({
    "customer_id":[1,2,1,3],
    "amount":[500,700,300,400]
})
merged = pd.merge(customers, orders, on="customer_id")
print(merged)
total_purchase = merged.groupby("customer_name")["amount"].sum()
print(total_purchase)
```
